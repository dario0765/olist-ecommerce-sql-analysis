```python
import pandas as pd
import pymysql
from sqlalchemy import create_engine
```

```python
print(f"pandas: {pd.__version__}")
print(f"sqlalchemy: {sqlalchemy.__version__}")
print(f"pymysql: {pymysql.__version__}")
```

```python
# Replace with your own local MySQL credentials
engine = create_engine("mysql+pymysql://YOUR_USER:YOUR_PASSWORD@localhost:3306/ecommerce_db")
```

```python
df_orders = pd.read_csv("olist_orders_dataset.csv")

print(f"Rows in CSV: {len(df_orders)}")

date_cols = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]
for col in date_cols:
    df_orders[col] = pd.to_datetime(df_orders[col], errors="coerce")

df_orders.to_sql("orders", con=engine, if_exists="append", index=False)
print(f"Loaded Rows: {len(df_orders)}")
```

```python
faltantes_df = pd.DataFrame({
    "product_category_name": ["pc_gamer", "portateis_cozinha_e_preparadores_de_alimentos"],
    "product_category_name_english": ["pc_gamer", "kitchen_portables_and_food_preparers"]
})

faltantes_df.to_sql("product_category_translation", con=engine, if_exists="append", index=False)
print("Categorías faltantes agregadas")
```

```python
df_products = pd.read_csv("olist_products_dataset.csv")
df_products.to_sql("products", con=engine, if_exists="append", index=False)
print(f"Loaded Rows: {len(df_products)}")
```

```python
df_order_items = pd.read_csv("olist_order_items_dataset.csv")

print(f"Rows in CSV: {len(df_order_items)}")

df_order_items["shipping_limit_date"] = pd.to_datetime(df_order_items["shipping_limit_date"], errors = "coerce")

df_order_items.to_sql("order_items", con=engine, if_exists="append", index=False)
print(f"Loaded Rows: {len(df_order_items)}")
```

```python
df_order_payments = pd.read_csv("olist_order_payments_dataset.csv")
df_order_payments.to_sql("order_payments", con=engine, if_exists="append", index=False)
print(f"Loaded Rows: {len(df_order_payments)}")
```
